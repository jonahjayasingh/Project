```python
from docx import Document

# Load the DOCX template file
template_path = "Doc1.docx"  # Replace with your file name
output_path = "internship_filled.docx"      # Output file name

# Dictionary of values to replace
data = {
    "{name}": "Jonah Jayasingh",
    "{date}": "1st June 2025 to 31st August 2025",
    "{coursename}": "Full Stack Web Development",
    "{projecttitle}": "Online Learning Management System"
}

# Load the document
doc = Document(template_path)

# Replace text in paragraphs
for paragraph in doc.paragraphs:
    for key, value in data.items():
        if key in paragraph.text:
            paragraph.text = paragraph.text.replace(key, value)

# Replace text in tables (if placeholders are inside tables)
for table in doc.tables:
    for row in table.rows:
        for cell in row.cells:
            for key, value in data.items():
                if key in cell.text:
                    cell.text = cell.text.replace(key, value)

# Save the new filled document
doc.save(output_path)

print(f"Internship certificate saved as: {output_path}")
```
