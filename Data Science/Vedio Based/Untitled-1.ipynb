```python
import cv2
import pytesseract
from moviepy import VideoFileClip
import speech_recognition as sr
import os
from tqdm import tqdm
```

```python
# ----------------------------
# 1️⃣ Extract text from video frames (OCR)
# ----------------------------

video_path = "demo.mp4"  # Replace with your video file
cap = cv2.VideoCapture(video_path)
```

```python
# Get total number of frames for progress bar
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

frame_number = 0

print("Extracting text from video frames...\n")
with tqdm(total=total_frames, desc="Frames processed") as pbar:
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        # Optional preprocessing for better OCR
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # OCR
        text = pytesseract.image_to_string(gray)
        if text.strip():  # only print non-empty text
            print(f"[Frame {frame_number}] {text.strip()}")
        
        frame_number += 1
        pbar.update(1)

cap.release()
```
