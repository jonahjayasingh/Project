```python
from ultralytics import YOLO

# Load a YOLOv8 model (small version; you can also use 'yolov8m.pt', 'yolov8l.pt', etc.)
model = YOLO('yolov8s.pt')  # This uses pretrained weights as a starting point

# Train the model
model.train(
    data='Indian-Traffic-Sign-1/data.yaml',  # Path to your dataset YAML
    epochs=10,
    imgsz=320,
    batch=8,
    name='indian-traffic-sign',
)

model = YOLO('runs/detect/indian-traffic-sign/weights/best.pt')

# Validate on the validation set
metrics = model.val()  # Uses validation split defined in data.yaml
print("Validation metrics:", metrics)

# Optional: Evaluate on test set (if defined in data.yaml)
metrics_test = model.val(split='test')  # Only works if 'test' is defined in YAML
print("Test metrics:", metrics_test)

# Predict on a single image (you can use a directory or video as source too)
results = model.predict(source='/Volumes/CrucialX9/Project/Data Science/Image Based/Traffic sign detection/Indian-Traffic-Sign-1/train/images/00003_jpg.rf.0fcfbbe4c3f56b4e311dd5ae91ce6c6b.jpg', conf=0.25, save=True)

# Optionally, view predictions
for r in results:
    print(r.boxes.xyxy)  # Bounding box coordinates
    print(r.boxes.cls)   # Predicted class indices
    print(r.boxes.conf)  # Confidence scores
```
