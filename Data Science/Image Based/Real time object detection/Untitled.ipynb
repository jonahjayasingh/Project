```python
import cv2
from ultralytics import YOLO

# Use YOLOv8-small for CPU
model = YOLO("yolov8n.pt")

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model.predict(source=frame, conf=0.5, device="mps")
    frame = results[0].plot()

    cv2.imshow("Real-Time Detection (CPU)", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
